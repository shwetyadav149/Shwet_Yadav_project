```python
job_role = input("Enter Desired Job Role:")

prompt = f"""
you are an expert technical interviewer.

Generate ONLY ONE professional interview question for the role of:

{job_role}

Make it realistic and suitable for real interviews.
Do Not provide answer.
"""

response = llm.invoke(prompt)

question= response.content

print(question)

candidate_answer = input("Enter Your Answer:")

evaluation_prompt = f"""
You are an expert Interview evaluator.

Interview Question:
{question}

Candidate Answer:
{candidate_answer}

Evaluate ONLY based on this question-answer pair.

Provide:
1.Score out of 10
2.Strengths
3.Weaknesses
4.Improvement Suggestions
"""

evaluation_response = llm.invoke(evaluation_prompt)

print(evaluation_response.content)


```

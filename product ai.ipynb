```python
import os
import requests
import mysql.connector
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from google import genai
```

```python
load_dotenv()

GROQ_API_KEY = os.getenv("GROQ_API_KEY")
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")

print("Groq Key:", GROQ_API_KEY)
print("GEMINI Key:", GEMINI_API_KEY)
```

```python
db = mysql.connector.connect(
    host=os.getenv("DB_HOST"),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD"),
    database=os.getenv("DB_NAME")
)

cursor = db.cursor()

print("Database Connected ✅")
```

```python
llm = ChatGroq(
    groq_api_key=GROQ_API_KEY,
    model="llama-3.1-8b-instant"
)

print("LLM Ready ✅")
```

```python
client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

response = client.models.generate_content(
    model="gemini-2.5-flash",   # use this (stable & fast)
    contents="Hello, how are you?"
)

print(response.text)
```

```python


def search_product(query):
    prompt = f"""
    User searched for: {query}

    Show top 5 products available in India from platforms like Amazon, Flipkart, Meesho.
    Include:
    - Product name
    - Estimated price
    - Platform (Amazon/Flipkart/Meesho)
    - Short comparison

    Keep it realistic.
    """

    gemini_res = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    ).text

    final_res = llm.invoke(f"Format this nicely:\n{gemini_res}")

    return final_res.content


# 🔁 Continuous loop
while True:
    query = input("\nEnter product (or type 'exit' to quit): ")

    if query.lower() == "exit":
        print("Exiting... 👋")
        break

    result = search_product(query)

    print("\n🔍 Results:\n")
    print(result)
```
